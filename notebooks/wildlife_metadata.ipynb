```python
merged_df = annotation_df.merge(
    category_df,
    left_on='category_id',
    right_on='id',
    how='left'
)
```

```python
merged_df.head()
```

```python
merged_df = merged_df.drop(columns=['id_x', 'id_y'], axis=1)
```

```python
merged_df = merged_df.rename(columns={'name': 'species'})
```

```python
merged_df.head()
```

```python
df = merged_df.merge(
    image_df,
    left_on='image_id',
    right_on='id',
    how='left'
)
```

```python
df
```

```python
df = df.drop(columns=['id', 'file_name'], axis=1)
```

```python
df
```

```python
df['datetime'].value_counts().head()
```

```python
common_names = {
    "oryx gazella": "Oryx/Gemsbok",
    "equus zebra hartmannae": "Hartmann's mountain zebra",
    "hyaena brunnea": "Brown hyena",
    "struthio camelus": "Ostrich",
    "antidorcas marsupialis": "Springbok",
    "giraffa camelopardalis": "Giraffe",
    "diceros bicornis": "Black rhino",
    "panthera leo": "Lion",
    "canis mesomelas": "Black-backed jackal",
    "loxodanta africana": "Elephant",
}
```

```python
subset_df = df[df['species'].isin(common_names.keys())].copy()
subset_df['common_name'] = subset_df['species'].map(common_names)
```

```python
SOFT_CAP = 2000  # bounds training time, but doesn't force equal counts

capped_df = (
    subset_df.groupby("species", group_keys=False)
    .apply(lambda x: x.sample(n=min(len(x), SOFT_CAP), random_state=42))
)
```

```python
capped_df
```

```python
capped_df.to_csv('manifest_metadata.csv', index=False)
```
